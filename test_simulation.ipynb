```python
# from game_theory_simulations import run_full_pipeline

from game_theory_simulations.trait_generator import extract_traits_from_game
```

```python
game_description = """
The Prisoner's Dilemma involves two players who must choose
to cooperate or defect. If both cooperate, both get a moderate
reward. If both defect, both get a small penalty. If one defects
while the other cooperates, the defector gets the best outcome
while the cooperator gets the worst.
"""
```

```python
game_traits = extract_traits_from_game(game_description, verbose=True)
```

```python
game_name = game_traits.game_name
from game_theory_simulations.config import sanitize_name
game_traits.game_name = sanitize_name(game_name)

# Save game configuration
game_traits.save()
```

```python
from game_theory_simulations.trait_generator import generate_trait_artifacts
```

```python
artifact_paths = generate_trait_artifacts(game_traits, verbose=True)

result = {
    "game_traits": game_traits,
    "artifact_paths": artifact_paths,
}
```

```python
from game_theory_simulations.vector_pipeline import generate_vectors_for_game

vector_paths = generate_vectors_for_game(
    game_name=game_traits.game_name,
    verbose=True,
)

result["vector_paths"] = vector_paths
```

```python
# Generate everything from a game description
result = run_full_pipeline(
    game_description="""
    The Prisoner's Dilemma involves two players who must choose
    to cooperate or defect. If both cooperate, both get a moderate
    reward. If both defect, both get a small penalty. If one defects
    while the other cooperates, the defector gets the best outcome
    while the cooperator gets the worst.
    """,
    game_name="prisoners_dilemma",
    generate_vectors=True,  # Set False to skip GPU-intensive step
)
```

```python
# Step 1: Extract traits from a game description
python game_theory_simulations/trait_generator.py \
    --game prisoners_dilemma  # Use a preset game
    # OR
    --description "Your custom game description..."

# Step 2: Generate persona vectors (requires GPU)
python game_theory_simulations/vector_pipeline.py \
    --game prisoners_dilemma

# Step 3: Launch the interactive app
streamlit run game_theory_simulations/app.py
# OR
python game_theory_simulations/run_app.py
```
